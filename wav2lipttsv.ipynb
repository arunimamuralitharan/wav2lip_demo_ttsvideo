```python
from gtts import gTTS

def text_to_video(text, face_path, output_path="/content/output.mp4"):
    # Step 1: TTS
    gTTS(text=text, lang='en').save("/content/speech.mp3")
    os.system("ffmpeg -y -i /content/speech.mp3 -ar 16000 -ac 1 /content/speech.wav")

    # Step 2: Lipsync
    os.system(f"""
        python /content/Wav2Lip/inference.py \
        --checkpoint_path /content/Wav2Lip/checkpoints/wav2lip_gan.pth \
        --face /content/face.jpg \
        --audio /content/speech.wav \
        --outfile /content/output.mp4 \
        --resize_factor 1
    """)
    print(f"Done! Video saved to /content/output.mp4")

# Usage
text_to_video("Hello world, this is my AI video pipeline.", "/content/face.mp4")
```
